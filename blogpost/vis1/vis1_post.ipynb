1/ Pyramide 
This first visualization focuses on representing all the names at the same time. All the names are ranked in brackets depending on their « popularity percentage » (that is, given a year, the percentage of new-born who have this name), along the y-axis. The x-axis is not significant, it just allows to display the different names in the space. 
Consequently, the strength of our visualization is to represent the entire set of names and the evolution of the baby name distribution along the years (since a slider allows to display the distribution over a chosen year, with a desired effect of continuity). Thus, the visualization allows to find out global trends over time (such as the diversification of the names over time). Also, one of our goal was to distinguish the more popular names from the rarest, while making these rare names as visible as the popular ones (which, for instance, is not the case of space filling approaches). We achieved this through encoding the popularity through verticality. Moreover, one can also see how one precise name evolves over time thanks to the selection option. 
However, our visualization is quite counter-intuitive. At first glance, one could think the dots do not represent names, but people. Moreover, the display of group of names in columns do not add anything to the visualization and may bring confusion. 


PS : a first idea that we tried was inspired by the short videos that rank different entities over time (eg this video). Because of the lack of animation tools offered by Altair, we don’t get a smooth continuity effect as desire.
